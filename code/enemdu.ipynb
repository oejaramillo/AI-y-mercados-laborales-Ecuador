```python
import numpy as np
import pandas as pd
```

Queremos cruzar los datos de nuestro crosswalk con los datos de ENEMDU 2025 para ver cuales son las ocupaciones que tenemos

```python
enemdu = pd.read_spss(r'C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\ENEMDU\1_BDD_ENEMDU_2025_SPSS\BDDenemdu_personas_2025_anual.sav', convert_categoricals=False)
```

```python
cr = pd.read_csv(r'C:\edu\repositories\AI-y-mercados-laborales-Ecuador\crosswalk\cw-own.csv', index_col=0)
```

```python
cr['Código CIUO'] = cr['Código CIUO'].apply(lambda x: str(x).zfill(4))
enemdu['p40'] = enemdu['p40'].apply(lambda x: str(int(float(x))).zfill(4) if pd.notna(x) else np.nan)
```

```python
merge = pd.merge(enemdu, cr, left_on='p40', right_on='Código CIUO', how='left')
```

```python
merge.to_parquet(r'C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\enemdu_con_tareas.parquet')
```

```python
pd.Series(enemdu['p40'].unique()).to_csv(r'C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\ocupaciones_unicas_ecuador.csv')
```
