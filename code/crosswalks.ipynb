## Crosswalk BLS

Queremos primero utilizar el crosswalk de Bureau of Labor Statistics entre las ocupaciones O*NET (SOC) e ISCO-08 que se corresponde con la clasificación CIIUO de la ENEMDU 

```python
data = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\crosswalk\crosswalkBLS.csv", sep="\t", dtype=str)
```

```python
onet_labels = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\full_onet_data.tsv", sep="\t")
onet_full = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\full_labelset.tsv", sep='\t')
```

```python
onet_full['O*NET-SOC Code'] = onet_full['O*NET-SOC Code'].apply(lambda x: x[:-3])
```

```python
data['2010 SOC Code'] = data['2010 SOC Code'].apply(lambda x: x.strip())
```

```python
crw = pd.merge(data, onet_full, left_on="2010 SOC Code", right_on="O*NET-SOC Code", how="outer")
```

```python
crw[['ISCO-08 Code', 'ISCO-08 Title EN', '2010 SOC Code', '2010 SOC Title', 'Task ID', 'Task', 'Task Type', 'Title',
       'human_exposure_agg', 'gpt4_exposure', 'gpt4_exposure_alt_rubric',
       'gpt_3_relevant', 'gpt4_automation', 'alpha', 'beta', 'gamma', 'automation', 'human_labels']].to_csv(
           r'C:\edu\repositories\AI-y-mercados-laborales-Ecuador\crosswalk\cw-bls.csv')
```

## Crosswalk Own labels

```python
data = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\crosswalk\crosswalkOwn.csv", sep=',', encoding='latin1', dtype=str)
bls = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\crosswalk\crosswalkBLS.csv", sep="\t", dtype=str)
```

```python
cr = pd.merge(data, bls, left_on='ocupaciones ONET', right_on='2010 SOC Title', how='left')
```

```python
onet_labels = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\full_onet_data.tsv", sep="\t")
onet_full = pd.read_csv(r"C:\edu\repositories\AI-y-mercados-laborales-Ecuador\data\full_labelset.tsv", sep='\t')
```

```python
cr['Código CIUO'] = cr['Código CIUO'].apply(lambda x: str(x).replace(',', '.').split('.')[0].zfill(4))
```

```python
onet_full['O*NET-SOC Code'] = onet_full['O*NET-SOC Code'].apply(lambda x: str(x).replace('.', '-').split('-')[1].zfill(4))
cr['2010 SOC Code'] = cr['2010 SOC Code'].apply(lambda x: str(x).strip().split('-')[-1].zfill(4) if pd.notna(x) else np.nan)
```

```python
crw = pd.merge(cr, onet_full, left_on="2010 SOC Code", right_on="O*NET-SOC Code", how="inner")
```

```python
crw[['ocupaciones ONET', 'Código CIUO', 'ISCO-08 Code', 'ISCO-08 Title EN', '2010 SOC Code', '2010 SOC Title', 'Task ID', 'Task', 'Task Type', 'Title',
       'human_exposure_agg', 'gpt4_exposure', 'gpt4_exposure_alt_rubric',
       'gpt_3_relevant', 'gpt4_automation', 'alpha', 'beta', 'gamma', 'automation', 'human_labels']].to_csv(
           r'C:\edu\repositories\AI-y-mercados-laborales-Ecuador\crosswalk\cw-own.csv')
```
